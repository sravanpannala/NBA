# Finding largest increase in metrics from Reg Season to Playoffs

```python
import os, sys

sys.path.append(os.path.dirname(os.path.dirname(os.path.abspath("__file__"))))
from nbafuns import *

data_DIR = "../data/box/"
fig_DIR = "../figs/playoffs/"

player_dict = get_players_pbp(league="NBA")
img_DIR_P = "../data/images/players/"
```

## PPG

```python
df1 = get_box(PT="P",measure="Base",cum=True,seasons=range(2020,2025),playoffs=False)
df2 = get_box(PT="P",measure="Base",cum=True,seasons=range(2020,2025),playoffs=True)
df1_1 = df1.query("gp >= 20")
df3 = df1_1[["player_id","season","pts"]]
df2_1 = df2.query("gp >= 4")
df4 = df2_1[["player_id","season","pts"]]
df5 = pd.merge(df3,df4,on=["player_id","season"])
df5["diff"] = round(df5["pts_y"] - df5["pts_x"],1)
df6 = df5.sort_values("diff",ascending=False).reset_index(drop=True)
df6["player"] = df6["player_id"].map(player_dict)
df6.index +=1
df6.insert(1,"player",df6.pop("player"))
df6["player_id"] = df6["player_id"].astype("str")
df6 = df6.reset_index()
df7 = df6.head(20)
```

```python
t = (
    GT(df7)
        .tab_header(
            title = md(f"**NBA Playoff Risers: Past 5 Seasons**"),
            subtitle = md("Largest Increase in **Points per Game** from Regular Season to Playoffs.<br />Minimum 20 games played in regular season and 4 games played in playoffs")  
        )
        .tab_source_note(source_note="socials:@sradjoker | source: nba.com/stats via nba_api")
        .cols_label(
            index = "#",
            player_id = "",
            player = "Player",
            pts_x = "Regular Season",
            pts_y="Playoffs",
            diff="Increase",
            season="Season",
        )
        .data_color(
            columns=['pts_x','pts_y'],
            palette="RdYlGn",
            # reverse="False"
        )
        .fmt_image(
            columns="player_id",
            path = img_DIR_P,
            file_pattern="{}.png"
        )
        .cols_align(align="center")
        .cols_align(align="left", columns="player")
        .tab_options(
            heading_title_font_size="200%",
            heading_subtitle_font_size="100%",
            # heading_title_font_weight='bold',
            table_background_color="floralwhite",
            column_labels_font_size="105%",
            column_labels_font_weight='bold',
            row_group_font_weight='bold',
            row_group_background_color="#E5E1D8",
            table_font_size=10,
            table_font_names="Consolas", 
            data_row_padding = "0.5px",
            # table_margin_left = 7,
            # table_margin_right = 0,
    )
)
t.save(fig_DIR + f"NBA_playoff_risers_PPG.png",scale=3,web_driver="firefox")
t
```

## DRE

```python
df1_1 = get_box(PT="P",measure="Base",cum=True,seasons=range(2020,2025),playoffs=False)
df1_2 = get_box(PT="P",measure="Adv",cum=True,seasons=range(2020,2025),playoffs=False)
df1_2 = df1_2[["player_id","poss","season"]]
df1 = pd.merge(df1_1,df1_2,on=['player_id','season'])
df1["poss"] = df1["poss"]/df1["gp"]
df2_1 = get_box(PT="P",measure="Base",cum=True,seasons=range(2020,2025),playoffs=True)
df2_2 = get_box(PT="P",measure="Adv",cum=True,seasons=range(2020,2025),playoffs=True)
df2_2 = df2_2[["player_id","poss","season"]]
df2 = pd.merge(df2_1,df2_2,on=['player_id','season'])
df2["poss"] = df2["poss"]/df2["gp"]
df1["dre"] = -8.42364*df1["poss"]/100+0.79231*df1["pts"]-0.71944*(df1["fga"]-df1["fg3a"])-0.55233*df1["fg3a"]-0.15944*df1["fta"]+0.13479*df1["oreb"]+0.3996*df1["dreb"]+0.54415*df1["ast"]+1.68007*df1["stl"]+0.76387*df1["blk"]-1.3599*df1["tov"]-0.10838*df1["pf"]
df2["dre"] = -8.42364*df2["poss"]/100+0.79231*df2["pts"]-0.71944*(df2["fga"]-df2["fg3a"])-0.55233*df2["fg3a"]-0.15944*df2["fta"]+0.13479*df2["oreb"]+0.3996*df2["dreb"]+0.54415*df2["ast"]+1.68007*df2["stl"]+0.76387*df2["blk"]-1.3599*df2["tov"]-0.10838*df2["pf"]
df1["dre"] = df1["dre"].round(2)
df2["dre"] = df2["dre"].round(2)
```

### Risers

```python
df1_3 = df1.query("gp >= 20")
df3 = df1_3[["player_id","season","dre"]]
df2_3 = df2.query("gp >= 4")
df4 = df2_3[["player_id","season","dre"]]
df5 = pd.merge(df3,df4,on=["player_id","season"])
df5["diff"] = round(df5["dre_y"] - df5["dre_x"],2)
df6 = df5.sort_values("diff",ascending=False).reset_index(drop=True)
df6["player"] = df6["player_id"].map(player_dict)
df6.index +=1
df6.insert(1,"player",df6.pop("player"))
df6["player_id"] = df6["player_id"].astype("str")
df6 = df6.reset_index()
df7 = df6.head(20)
t = (
    GT(df7)
        .tab_header(
            title = md(f"**NBA Playoff Risers: Past 5 Seasons**"),
            subtitle = md("Largest Increase in **DRE** from Regular Season to Playoffs.<br />Minimum 20 games played in regular season and 4 games played in playoffs.<br /> **DRE** (Daily RAPM Estimate) is a stat created by @NBACouchside similar to GameScore")  
        )
        .tab_source_note(source_note="socials:@sradjoker | source: nba.com/stats via nba_api")
        .cols_label(
            index = "#",
            player_id = "",
            player = "Player",
            dre_x = "Regular Season",
            dre_y="Playoffs",
            diff="Difference",
            season="Season",
        )
        .data_color(
            columns=['dre_x','dre_y'],
            palette="RdYlGn",
            # reverse="False"
        )
        .fmt_image(
            columns="player_id",
            path = img_DIR_P,
            file_pattern="{}.png"
        )
        .cols_align(align="center")
        .cols_align(align="left", columns="player")
        .tab_options(
            heading_title_font_size="200%",
            heading_subtitle_font_size="100%",
            # heading_title_font_weight='bold',
            table_background_color="floralwhite",
            column_labels_font_size="105%",
            column_labels_font_weight='bold',
            row_group_font_weight='bold',
            row_group_background_color="#E5E1D8",
            table_font_size=10,
            table_font_names="Consolas", 
            data_row_padding = "0.5px",
            # table_margin_left = 7,
            # table_margin_right = 0,
    )
)
t.save(fig_DIR + f"NBA_playoff_risers_DRE.png",scale=3,web_driver="firefox")
t
```

## 5 Year Average

```python
seasons = np.arange(2020,2025)
df1_1 = get_box(PT="P",measure="Base",cum=False,seasons=seasons,playoffs=False)
df1_1["game_id"] = df1_1["game_id"].astype(int)
df1_2 = get_box(PT="P",measure="Adv",cum=False,seasons=seasons,playoffs=False)
df1_2 = df1_2[["gameid","personid","possessions"]]
df1_2.columns = ['game_id','player_id',"poss"]
df1 = pd.merge(df1_1,df1_2,on=['player_id','game_id'])
df2_1 = get_box(PT="P",measure="Base",cum=False,seasons=seasons,playoffs=True)
df2_1["game_id"] = df2_1["game_id"].astype(int)
df2_2 = get_box(PT="P",measure="Adv",cum=False,seasons=seasons,playoffs=True)
df2_2 = df2_2[["gameid","personid","possessions"]]
df2_2.columns = ['game_id','player_id',"poss"]
df2 = pd.merge(df2_1,df2_2,on=['player_id','game_id'])
df1["dre"] = -8.42364*df1["poss"]/100+0.79231*df1["pts"]-0.71944*(df1["fga"]-df1["fg3a"])-0.55233*df1["fg3a"]-0.15944*df1["fta"]+0.13479*df1["oreb"]+0.3996*df1["dreb"]+0.54415*df1["ast"]+1.68007*df1["stl"]+0.76387*df1["blk"]-1.3599*df1["tov"]-0.10838*df1["pf"]
df2["dre"] = -8.42364*df2["poss"]/100+0.79231*df2["pts"]-0.71944*(df2["fga"]-df2["fg3a"])-0.55233*df2["fg3a"]-0.15944*df2["fta"]+0.13479*df2["oreb"]+0.3996*df2["dreb"]+0.54415*df2["ast"]+1.68007*df2["stl"]+0.76387*df2["blk"]-1.3599*df2["tov"]-0.10838*df2["pf"]
df1["dre"] = df1["dre"].round(2)
df2["dre"] = df2["dre"].round(2)
df1["pra"] = df1["pts"] + df1["reb"] + df1["ast"]
df2["pra"] = df2["pts"] + df2["reb"] + df2["ast"]
```

```python
df1.columns
```

```python
df1["mul"] = df1["dre"]*df1["poss"]
df1g = df1.groupby(["player_id"])[["poss","mul","pra"]].agg({"poss":["count","sum"],"mul":["sum"],"pra":["sum"]})
df1g = df1g.reset_index()
df1g.columns = ["player_id","gp","poss","mul","pra"]
df1g["dre"] = round(df1g["mul"]/df1g["poss"],2)
df1g = df1g[["player_id","gp","pra","dre"]]
df2["mul"] = df2["dre"]*df2["poss"]
df2g = df2.groupby(["player_id"])[["poss","mul","pra"]].agg({"poss":["count","sum"],"mul":["sum"],"pra":["sum"]})
df2g = df2g.reset_index()
df2g.columns = ["player_id","gp","poss","mul","pra"]
df2g["dre"] = round(df2g["mul"]/df2g["poss"],2)
df2g = df2g[["player_id","gp","pra","dre"]]
```

```python
df3 = pd.merge(df1g,df2g,on=["player_id"])
df3["player"] = df3["player_id"].map(player_dict)
df3.insert(1,"player",df3.pop("player"))
```

### Risers

```python
df4 = df3.query("gp_x>=82 & gp_y>=8")
df4["diff"] = round(df4["dre_y"] - df4["dre_x"],2)
df5 = df4.sort_values("diff",ascending=False).reset_index(drop=True)
df5.index +=1
df5 = df5.reset_index()
df5["player_id"] = df5["player_id"].astype("str")
df6 = df5.head(20)
t = (
    GT(df6)
        .tab_header(
            title = md(f"**NBA Playoff Risers: Past 5 Seasons Average**"),
            subtitle = md("Largest Increase in **DRE** from Regular Season to Playoffs.<br />Minimum 82 games played in regular season and 8 games played in playoffs over 5 seasons.<br /> **DRE** (Daily RAPM Estimate) is a stat created by @NBACouchside similar to GameScore")  
        )
        .tab_source_note(source_note="socials:@sradjoker | source: nba.com/stats via nba_api")
        .cols_label(
            index = "#",
            player_id = "",
            player = "Player",
            gp_x = "GP",
            dre_x = "DRE",
            gp_y = "GP",
            dre_y="DRE",
            diff="Difference",
        )
        .tab_spanner(
            label="Regular Season",
            columns=["gp_x","dre_x"],
        )
        .tab_spanner(
            label="Playoffs",
            columns=["gp_y","dre_y"],
        )
        .data_color(
            columns=['dre_x','dre_y'],
            palette="RdYlGn",
            # reverse="False"
        )
        .fmt_image(
            columns="player_id",
            path = img_DIR_P,
            file_pattern="{}.png"
        )
        .cols_align(align="center")
        .cols_align(align="left", columns="player")
        .tab_options(
            heading_title_font_size="150%",
            heading_subtitle_font_size="100%",
            # heading_title_font_weight='bold',
            table_background_color="floralwhite",
            column_labels_font_size="105%",
            column_labels_font_weight='bold',
            row_group_font_weight='bold',
            row_group_background_color="#E5E1D8",
            table_font_size=10,
            table_font_names="Consolas", 
            data_row_padding = "0.5px",
            # table_margin_left = 7,
            # table_margin_right = 0,
    )
)
t.save(fig_DIR + f"NBA_playoff_risers_avg_DRE.png",scale=3,web_driver="firefox")
t
```
