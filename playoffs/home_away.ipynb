```python
import os, sys

sys.path.append(os.path.dirname(os.path.dirname(os.path.abspath("__file__"))))
from nbafuns import *
```

```python
seasons = np.arange(1996,2025)
df0 = get_box(PT="T",measure="Base",cum=False,seasons=seasons,playoffs=True)
df0["home"] = ~df0["matchup"].str.contains("@")
df0["w"] = np.where(df0["wl"] == "W",True,False)
df1 = df0.query("home == True")
df2 = df1.groupby("season")[["w"]].value_counts()
df2 = df2.unstack().reset_index()
df2.columns = ["season","l","w"]
df2["w_pct"] = df2["w"]/(df2["w"]+df2["l"])
df2 = df2[df2["season"]!=2020]
```

```python
p = (
    ggplot(df2)
    + aes(x="season",y="w_pct")
    + geom_point(show_legend=False)
    + geom_line(size=1,show_legend=False)
    + geom_smooth(method="lm",color="red",se=False)
    + geom_hline(yintercept=0.6, color = "blue", linetype="dashed")
    + geom_hline(yintercept=0.5, color = "blue", linetype="dashed")
    + theme_idv
    + scale_y_continuous(labels=percent_format())
    + labs(
        title="Win % of Home Teams in Playoffs",
        subtitle="From 1996-97 NBA Season",
    )
    + pnba
)
p
```

```python
seasons = np.arange(1996,2025)
df0 = get_box(PT="T",measure="Base",cum=False,seasons=seasons,playoffs=True)
df0["home"] = ~df0["matchup"].str.contains("@")
df0["w"] = np.where(df0["wl"] == "W",True,False)
df1 = df0.sort_values("home").groupby("game_id")
# df2 = df1.groupby("season")[["w"]].value_counts()
# df2 = df2.unstack().reset_index()
# df2.columns = ["season","l","w"]
# df2["w_pct"] = df2["w"]/(df2["w"]+df2["l"])
```

```python
cols = ["game_id","team_id","home","w","season"]
df1_1 = df1.nth(1)
df1_1 = df1_1[cols]
df1_2 = df1.nth(0)
df1_2 = df1_2[["game_id","team_id","season"]]
# df2 = pd.merge(df1_1,df1_2,on="game_id")
df3 = get_box(PT="T",measure="Base",cum=True,seasons=seasons,playoffs=False)
df3 = df3[["team_id","season","plus_minus"]]
df3_1 = pd.merge(df1_1,df3,on=["team_id","season"])
df3_2 = pd.merge(df1_2,df3,on=["team_id","season"])
df4 = pd.merge(df3_1,df3_2,on=["game_id","season"])
df4["diff"] = df4["plus_minus_x"] - df4["plus_minus_y"]
df4["bett"] = df4["diff"] >0
```

```python
df4["bett_w"] = (df4["bett"] & df4["w"])
df4["bett_l"] = (df4["bett"] & ~df4["w"])
```
