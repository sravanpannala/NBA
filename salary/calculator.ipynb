```python
import os, sys

sys.path.append(os.path.dirname(os.path.dirname(os.path.abspath("__file__"))))
from nbafuns import *

from scipy.optimize import fsolve
```

```python
def salary_func(x: int,years: int=4, rise: int=8, init: int=0) -> int:
    s = 0
    for i in range(years):
        s+=((1.08)**i)*x
    return s-init

def salary_vec(s: int,years: int=4, rise: int=8):
    x = np.zeros(years)
    for i in range(years):
        x[i]=round(((1+rise/100)**i)*s,2)
    return x

def percent_cap(s: int, cap: int):
    pcap = round(s/cap*100,1)
    return pcap

def cap_percent(s,cap):
    sa = s/100*cap
    return sa
```

```python
salary = 150
years = 5
rise = 8
fsolve(salary_func,x0=25,args=(4,8,400))
```
