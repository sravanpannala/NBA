```python
import os, sys

sys.path.append(os.path.dirname(os.path.dirname(os.path.abspath("__file__"))))
from nbafuns import *

fig_DIR = "../figs/analysis/"
img_DIR_T = "../data/images/teams/"
box_DIR = "../data/box/"
```

```python
seasons = range(1996,2025)
df1 = get_box("T","Base","True",seasons,playoffs=False)
```

```python
df2 = df1[["team_id","team_name","w_pct","season"]]
```

```python
df3 = add_tinfo(df2,on="team_id")
df3 = df3[['team_name_x', 'w_pct','idConference','season']]
df3.columns = ['team_name', 'w_pct','conf','season']
```

```python
df4 = df3.groupby(["season","conf"])["w_pct"].mean()
df4 = df4.unstack().reset_index()
df4.columns = ["season","east","west"]
df4["wb"] = df4["west"] > df4["east"]
```
