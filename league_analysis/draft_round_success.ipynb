```python
dfaim = pd.read_csv(data_DIR + "NBA_AIM_FULL.csv")
```

```python
cols = ["player_id", "ewins","LEBRON_WinsAdded","Season"]
```

```python
df1 = dfaim[cols]
df1.fillna(0, inplace=True)
df2 = df1.groupby("player_id")[["ewins","LEBRON_WinsAdded"]].sum().reset_index()
df2["ewp"] = df2["ewins"] > 0
df2["lbp"] = df2["LEBRON_WinsAdded"] > 0
```
