```python
import numpy as np
import pandas as pd
from icecream import ic
from tqdm import tqdm
import time
from IPython.display import clear_output

import requests
from nba_api.stats.library.http import NBAStatsHTTP

from nba_api.stats.endpoints import (
    leaguegamelog,
    leaguedashplayerstats,
    leaguedashteamstats,
)

pd.options.mode.chained_assignment =  None

box_DIR = "../../data/box/"
shiny_DIR = "../../data/shiny/"
```

```python
def get_box_p_cum(seasons, measure="Base", n=32):
    if measure == "Advanced":
        per_mode = "Per100Possessions"
    else:
        per_mode = "PerGame"
    for season in tqdm(seasons):
        try:
            season_str = season + "-" + str(int(season) + 1)[-2:]
            stats = leaguedashplayerstats.LeagueDashPlayerStats(
                measure_type_detailed_defense=measure,
                per_mode_detailed=per_mode,
                season=season_str,
                season_type_all_star="Playoffs"
            )
            df1 = stats.get_data_frames()[0]
            df = df1.iloc[:, :n]
            if measure == "Advanced":
                measure1 = "Adv"
            else:
                 measure1 = measure
            df.to_parquet(box_DIR + "NBA_Box_P_Cum_" + measure1 + "_" + season + "_PS.parquet")
            time.sleep(0.6)
        except Exception as error:
            continue

def update_box_p_cum(seasons):
    print("Base")
    get_box_p_cum(seasons, measure="Base", n=32)
    print("Advanced")
    get_box_p_cum(seasons, measure="Advanced", n=43)
    print("Misc")
    get_box_p_cum(seasons, measure="Misc", n=23)
    print("Scoring")
    get_box_p_cum(seasons, measure="Scoring", n=29)
```

```python
season_start = 2024
season_end = 2025
seasons = np.arange(season_start, season_end, 1).astype(str)
update_box_p_cum(seasons)
```

```python
def get_box_t_cum(seasons, measure="Base", n=32):
    if measure == "Advanced":
        per_mode = "Per100Possessions"
    else:
        per_mode = "PerGame"
    for season in tqdm(seasons):
        try:
            season_str = season + "-" + str(int(season) + 1)[-2:]
            stats = leaguedashteamstats.LeagueDashTeamStats(
                measure_type_detailed_defense=measure,
                per_mode_detailed=per_mode,
                season=season_str,
                season_type_all_star="Playoffs"
            )
            df1 = stats.get_data_frames()[0]
            df = df1.iloc[:, :n]
            if measure == "Advanced":
                measure1 = "Adv"
            else:
                 measure1 = measure
            df.to_parquet(box_DIR + "NBA_Box_T_Cum_" + measure1 + "_" + season + "_PS.parquet")
            time.sleep(0.6)
        except Exception as error:
            continue

def update_box_t_cum(seasons):
    print("Base")
    get_box_t_cum(seasons, measure="Base", n=28)
    print("Advanced")
    get_box_t_cum(seasons, measure="Advanced", n=27)
    print("Misc")
    get_box_t_cum(seasons, measure="Misc", n=15)
    print("Scoring")
    get_box_t_cum(seasons, measure="Scoring", n=22)
```
