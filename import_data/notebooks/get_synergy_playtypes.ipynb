```python
import time
from itertools import product

import numpy as np
import pandas as pd
from nba_api.stats.endpoints import synergyplaytypes
from tqdm import tqdm

export_DIR = "../../data/synergy/"
```

```python
playtypes = ["Transition","Isolation","PRBallHandler","PRRollman","Postup","Spotup","Handoff","Cut","OffScreen","OffRebound","Misc"]
typegrouping = ["offensive","defensive"]
```

```python
season_start = 2024
season_end = 2014
seasons = np.arange(season_start, season_end, -1)
```

```python
dfl = []
for season in tqdm(seasons):
    season_str = str(season)+"-"+str(season+1)[-2:]    
    dfa = []
    for typeg,typep in product(typegrouping,playtypes):
        # print(typeg,typep)
        for ii in range(5):
            try:
                stats = synergyplaytypes.SynergyPlayTypes(
                    league_id="00",
                    per_mode_simple="PerGame",
                    player_or_team_abbreviation="T",
                    season_type_all_star="Regular Season",
                    season=season_str,
                    play_type_nullable=typep,
                    type_grouping_nullable=typeg,
                )
                df1 = stats.get_data_frames()[0]
                dfa.append(df1)
                time.sleep(0.6)
                break
            except Exception as error:
                print(season)
                print(error)
                time.sleep(0.6)
                continue
    df = pd.concat(dfa)
    df.to_parquet(export_DIR +f"NBA_Synergy_T_Cum_{season}.parquet")
    dfl.append(df)
```
